```python
import pandas as pd
import sqlite3
```

```python
df = pd.read_csv(
    "data/Sample - Superstore.csv",
    encoding='latin1'
)
```

```python
conn = sqlite3.connect("database/superstore.db")
```

```python
df.to_sql(
    "superstore",
    conn,
    if_exists="replace",
    index=False
)
```

```python
pd.read_sql(
    "SELECT name FROM sqlite_master WHERE type='table';",
    conn
)
```

```python
#1.Total sales by category
query = """
SELECT Category, SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY Category;
"""

result = pd.read_sql(query, conn)
print(result)
```

```python
#2.Top 5 products by profit
query = """
SELECT "Product Name",
       SUM(Profit) AS Total_Profit
FROM superstore
GROUP BY "Product Name"
ORDER BY Total_Profit DESC
LIMIT 5;
"""

pd.read_sql(query, conn)
```

```python
#3.Sales by region
query = """
SELECT Region,
       SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY Region
ORDER BY Total_Sales DESC;
"""

pd.read_sql(query, conn)
```

```python
#4.Profit by category
query = """
SELECT Category,
       SUM(Profit) AS Total_Profit
FROM superstore
GROUP BY Category
ORDER BY Total_Profit DESC;
"""

pd.read_sql(query, conn)
```

```python
#5.Customer segment by spending
query = """
SELECT Segment,
       SUM(Sales) AS Total_Spending
FROM superstore
GROUP BY Segment
ORDER BY Total_Spending DESC;
"""

pd.read_sql(query, conn)
```

```python
#6.Most Sold Sub-Categories
query = """
SELECT "Sub-Category",
       SUM(Quantity) AS Total_Quantity
FROM superstore
GROUP BY "Sub-Category"
ORDER BY Total_Quantity DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
#7.Top 10 states by sales
query = """
SELECT State,
       SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY State
ORDER BY Total_Sales DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
#8.Avg discount by category
query = """
SELECT Category,
       AVG(Discount) AS Avg_Discount
FROM superstore
GROUP BY Category;
"""

pd.read_sql(query, conn)
```

```python
# 9.Technology Products (WHERE Clause)

query = """
SELECT *
FROM superstore
WHERE Category='Technology'
LIMIT 5;
"""

pd.read_sql(query, conn)
```

```python
#10.Products with sales above Average
query = """
SELECT "Product Name", Sales
FROM superstore
WHERE Sales >
(
    SELECT AVG(Sales)
    FROM superstore
)
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
#11.Rank Products by Sales
query = """
SELECT
    "Product Name",
    Sales,

    RANK() OVER
    (
        ORDER BY Sales DESC
    ) AS Sales_Rank

FROM superstore
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
#12.category Sales using CTE Requirement
query = """
WITH category_sales AS
(
    SELECT
        Category,
        SUM(Sales) AS Total_Sales
    FROM superstore
    GROUP BY Category
)

SELECT *
FROM category_sales;
"""

pd.read_sql(query, conn)
```

```python
# 13.Categories with Sales Greater than 500000 (HAVING Clause)

query = """
SELECT
    Category,
    SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY Category
HAVING Total_Sales > 500000;
"""

pd.read_sql(query, conn)
```
